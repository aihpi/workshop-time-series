<h3>2. Loading a Dataset</h3>
</div>

All course datasets are stored in Parquet format and their paths are defined in `nb_config.py`. Once a dataset has been prepared, loading it in any notebook takes two lines:

```python
import pandas as pd
from nb_config import OPS_15M_PATH

df = pd.read_parquet(OPS_15M_PATH)
df.head()
```

The available path constants are:

| Constant | Dataset |
|----------|---------|
| `OPS_15M_PATH` | OPS electricity consumption, 15-minute frequency |
| `OPS_30M_PATH` | OPS electricity consumption, 30-minute frequency |
| `CDC_TEMP_PATH` | CDC regional air temperature for Germany, monthly |
