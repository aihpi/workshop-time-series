<h3 id="3.-Environment-Setup">3. Environment Setup</h3>
</div>

This course uses **[uv](https://docs.astral.sh/uv/)**, a fast Python package and project manager. It pins every dependency to an exact version in `uv.lock`, so everyone working through the notebooks runs the same environment regardless of their machine.

> **Why not conda or plain pip?** Both work, but they do not guarantee identical environments across different machines. `uv` is also significantly faster to install and resolve packages than either of them.

---

#### Step 1: Install uv

Run one of the following commands in your terminal depending on your operating system. You only need to do this once.

**macOS and Linux**
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

**Windows (PowerShell)**
```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

After installation, restart your terminal and confirm it worked:
```bash
uv --version
```

> **Alternative:** If you would rather not run an install script, you can bootstrap with `pip install uv` and continue from the next step. You get the same tool either way.

---

#### Step 2: Clone the repository

```bash
git clone https://github.com/aihpi/kisz-time-series.git
cd kisz-time-series
```

> **Tip:** Course materials are released gradually throughout the course. We recommend working on a separate branch so that pulling updates does not conflict with your own work:
> ```bash
> git checkout -b my-work
> ```

---

#### Step 3: Sync the environment

From the repository root, run:

```bash
uv sync
```

This reads `pyproject.toml` and `uv.lock`, creates a virtual environment under `.venv/`, and installs all dependencies at the pinned versions. It is the equivalent of `pip install -r requirements.txt`, but reproducible.

---

#### Step 4: Open the project in VS Code

This course assumes you work in **[Visual Studio Code](https://code.visualstudio.com/)**. It runs
notebooks directly, and it picks up the environment `uv` just created without any further setup.

1. Install VS Code, if you do not have it yet.
2. Install the two extensions we need. Open the Extensions view (`Ctrl+Shift+X`, or `Cmd+Shift+X` on
   macOS) and install:
   - **Python** (`ms-python.python`)
   - **Jupyter** (`ms-toolsai.jupyter`)
3. Open the repository folder: *File > Open Folder...*, and select the folder you cloned in Step 2.
4. Open `notebooks/00_Practical_Introduction.ipynb` and click **Select Kernel** in the top-right
   corner of the notebook. Choose *Python Environments...* and pick the one marked `.venv`, the one
   inside the repository folder. VS Code remembers this choice for every notebook in the project.

> **Something looks off?** If `.venv` does not show up in the list, VS Code was probably opened on
> the wrong folder. It has to be the repository root, the folder that contains `pyproject.toml`.

> **Prefer JupyterLab?** It works just as well, it is simply not part of the environment we ship.
> Add it once with `uv add jupyterlab` and start it with `uv run jupyter lab`. Started that way, its
> default kernel is already the project environment. Everything else in the course stays the same.

---

#### uv commands you will use regularly

| Task | Command |
|------|---------|
| Install or sync all dependencies | `uv sync` |
| Add a new package | `uv add package_name` |
| Remove a package | `uv remove package_name` |
| Run a script inside the project environment | `uv run script.py` |
| List installed packages | `uv pip list` |
| Upgrade uv itself | `uv self update` |

Use `uv run` instead of `python` and `uv add` instead of `pip install` to keep the lock file consistent for everyone on the course.

<h3 id="4.-Verify-Your-Installation">4. Verify Your Installation</h3>
</div>

Run the cell below. Every package should show a green tick. If something shows a red cross, go back to your terminal, run `uv sync`, and then restart the kernel with the **Restart** button in the notebook toolbar.

````python
import importlib
import sys

# Packages required across the full course
required = {
    "pandas":      "Data loading and manipulation",
    "numpy":       "Numerical computing",
    "matplotlib":  "Plotting",
    "seaborn":     "Statistical visualisation",
    "statsmodels": "Statistical models (ARIMA, ETS, ...)",
    "sklearn":     "Machine learning utilities",
    "pyarrow":     "Parquet file support",
    "IPython":     "Notebook display utilities",
}

all_ok = True
for pkg, description in required.items():
    found = importlib.util.find_spec(pkg) is not None
    status = "✅" if found else "❌ NOT FOUND"
    if not found:
        all_ok = False
    print(f"{status}  {pkg:<15}  {description}")

print()
print(f"Python {sys.version}")
if all_ok:
    print("\n✅  All packages found. You are good to go!")
else:
    print("\n❌  Some packages are missing. Run 'uv sync' in the terminal, then restart the kernel.")
````
